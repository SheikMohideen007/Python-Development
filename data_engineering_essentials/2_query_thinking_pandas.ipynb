## **Fixing Data Types**

````python
import pandas as pd

# --------------------------------------------------
# 1. THE PROBLEM: Storing dates and numbers as strings
# --------------------------------------------------
print("--- THE PROBLEM (Strings) ---")

# Notice that February doesn't have a leading zero, and prices are quoted strings
bad_data = {
    'date': ["2025-11-01", "2025-2-01", "2025-12-01"],
    'price': ["10.50", "5.00", "abc"]
}
df_bad = pd.DataFrame(bad_data)
print("Original Data types:\n", df_bad.dtypes)

# Chronological sorting FAILS because it sorts alphabetically
# "2025-11-01" will incorrectly come BEFORE "2025-2-01"
df_sorted_bad = df_bad.sort_values(by='date')
print("\nIncorrectly Sorted (Alphabetical):")
print(df_sorted_bad)


average = df_bad['price'].mean()


````

````python
# --------------------------------------------------
# 2. THE SOLUTION: Converting to proper data types
# --------------------------------------------------
print("--- THE SOLUTION (Proper Types) ---")

df_good = df_bad.copy()

# Convert 'date' string to an actual datetime object
df_good['date'] = pd.to_datetime(df_good['date'])

# Convert 'price' string to a float
df_good['price'] = pd.to_numeric(df_good['price'], errors = "coerce")

print("Fixed Data types:\n", df_good.dtypes)

# Chronological sorting WORKS perfectly now
df_sorted_good = df_good.sort_values(by='date')
print("\nCorrectly Sorted (Chronological):")
print(df_sorted_good)

# Numeric operations WORK perfectly now
average_price = df_good['price'].mean()
print(f"\n✅ Success! The average price is: ${average_price:.2f}")
print(df_good)
````

````python
import pandas as pd

# 1. Create a highly realistic, messy user-input dataset
df = pd.DataFrame({
    'user_id': ['USR_01', 'USR_02', 'USR_03', 'USR_04', 'USR_05', 'USR_06'],
    # Notice the leading spaces, random capitalizations, and bad typos
    'gender_raw': ['male', '  m  ', 'female', 'f', 'maale', 'FEMALE']
})

print("📋 1. RAW MESSY INPUTS:")
print(df)
print("\n" + "="*50 + "\n")
````

````python
# 2. STEP ONE: Normalize the text first (Strip spaces and make lowercase)
# This instantly squashes "  m  " to "m" and "FEMALE" to "female"
df['gender_normalized'] = df['gender_raw'].astype(str).str.strip().str.lower()

print("🧹 2. AFTER COARSE CLEANING (.str.strip().str.lower()):")
print(df)
print("\n" + "="*50 + "\n")


````

````python
# 3. STEP TWO: Define a clean, uniform translation dictionary
# Since everything is now lowercase and stripped, our dictionary can be tiny!
clean_map = {
    'male': 'Male',
    'm': 'Male',
    'female': 'Female',
    'f': 'Female'
}


# 4. STEP THREE: Apply .replace()
# 'maale' is not in our dictionary keys, so .replace() leaves it as 'maale' safely.
df['gender_final'] = df['gender_normalized'].replace(clean_map)

print("✅ 3. FINAL OUTPUT MATRIX USING .replace():")
print(df[['user_id', 'gender_raw', 'gender_final']])
````

## Case Study: Analyzing Food Delivery Performance

We have been given a core business question from our operations team:
> **"Which city had the highest average order value last month, among orders that were not cancelled?"**

To solve this, we must break this human English request down into logical data steps.

### The 3-Step Analytical Pipeline:
1. **Filter:** Kick out any orders that were cancelled (and focus on last month's timeframe).
2. **Aggregate:** Group the remaining orders by `city` and calculate the average (`AVG` / `mean`) order cost.
3. **Sort & Extract:** Sort the list from highest average cost to lowest, and take the top result.

````python
import pandas as pd

# Load the Superstore-style dataset
data = {
    'order_id': ['O1','O2','O3','O4','O5','O6','O7','O8'],
    'city': ['Mumbai','Delhi','Mumbai','Bangalore','Delhi','Mumbai','Delhi','Bangalore'],
    'category': ['Tech','Furniture','Tech','Office','Tech','Furniture','Office','Tech'],
    'sales': [25000, 8000, 15000, 3000, 32000, 12000, 5500, 18000],
    'profit': [5000, -500, 3200, 800, 8000, 1200, 600, 4000],
    'status': ['completed','completed','completed','cancelled','completed','completed','cancelled','completed']
}
df = pd.DataFrame(data)
print(df)
````

````python
df[['order_id', 'status', 'profit']]
````

````python
df[df['sales'] > 10000][['order_id', 'sales']]

# select order_id, sales from data where sales > 10000
````
