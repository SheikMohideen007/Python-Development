# Understanding `iloc` and `loc` in Pandas

When working with DataFrames, we often need to select specific rows and columns.

Pandas provides two important selection methods:

## 1. iloc

`iloc` stands for **Integer Location**.

It selects data based on **position numbers**.

Think of it as:

- First row → Position 0
- Second row → Position 1
- Third row → Position 2

---

## 2. loc

`loc` stands for **Location using Labels**.

It selects data based on **row labels (index values)**.

Think of it as:

- Student Roll Number
- Customer ID
- Transaction ID

Instead of position, we use labels.

---

Let's understand with examples.

````python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Venkatesh", "Rahul", "Priya", "Ankit"],
    "Age": [28, 25, 30, 27],
    "Salary": [100000, 80000, 120000, 90000]
})

df
````

## Selecting a Single Row using `iloc`

Syntax:

```python
df.iloc[row_position]

Here, 0 means:

First row
Position 0

Question:

What do you think will be returned?

````python
# Code Cell

df.iloc[0]
````

## Selecting Multiple Rows using `iloc`

Python slicing rules apply.

Syntax:

```python
df.iloc[start:end]

````python
df.iloc[0:3]
````

returns:

Row 0
Row 1
Row 2

Row 3 is NOT included.

## Selecting Specific Rows and Columns using `iloc`

Syntax:

```python
df.iloc[row_range, column_range]

````python
df.iloc[0:3, 0:2]
````

Meaning:

Rows:

0
1
2

Columns:

0
1

Notice that the ending index is excluded for both rows and columns.

# The Three Common Strategies

When we find missing values, we usually choose one of three approaches.

Strategy 1: Drop

Remove rows or columns.

Strategy 2: Fill with a Constant

Examples:

Age = 0

City = "Unknown"

Strategy 3: Impute

Estimate a reasonable value using existing data.

Examples:

Mean

Median

Mode

# Important Beginner Confusion

Many students confuse:

dropna()

and

drop(columns=[...])

These are completely different.

dropna()

    Removes ROWS

drop(columns=[...])

    Removes COLUMNS

Notice:

Cabin is missing for most passengers.

Question:

If 77% of Cabin values are missing,
is it worth keeping?

Usually no.

This is why many Titanic analyses remove Cabin.
